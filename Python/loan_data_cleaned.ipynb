Step 6: Fix the age column

```python
age_mapping = {
    '<25' : 20,
    '25-34' : 30,
    '35-44' : 40,
    '45-54' : 50,
    '55-64' : 60,
    '65-74' : 70,
    '>74' : 78
}
```

```python
df['Age_Estimate'] = df['age'].map(age_mapping)
```

```python
df['Age_Estimate'].isnull().sum()
```

```python
df['Age_Estimate'].isnull().sum()
```

Step 7: Handle Outliers in LTV

```python
print("Rows with LTV above 100:", (df['LTV']>100).sum())
```

```python
# We cap (limit) LTV at 100, since LTV above 100% is not realistic for most standard loans
df['LTV'] = df['LTV'].apply(lambda x: 100 if x > 100 else x)
```

Step 8: Rename Status to Clear Labels

```python
df['Default_Status'] = df['Status'].map({1: 'Yes', 0: 'No'})
```

Step 9: Create Extra Realistic Columns

```python
# 1. EMI - calculated using a simplified loan EMI formula
df['Interest_Rate_Monthly'] = df['rate_of_interest'] / 12 / 100
df['EMI'] = (df['loan_amount'] * df['Interest_Rate_Monthly'] * 
             (1 + df['Interest_Rate_Monthly'])**df['term']) / \
            ((1 + df['Interest_Rate_Monthly'])**df['term'] - 1)

# 2. Branch - randomly assign one of 10 branch codes (simulating real bank branches)
np.random.seed(42)
branch_list = ['BR001','BR002','BR003','BR004','BR005','BR006','BR007','BR008','BR009','BR010']
df['Branch'] = np.random.choice(branch_list, size=len(df))

# 3. Loss_Amount - only applies to defaulted loans, estimated as a portion of loan amount
df['Loss_Amount'] = np.where(df['Default_Status'] == 'Yes', 
                              df['loan_amount'] * 0.35, 0)

# 4. Recovery_Amount - only applies to defaulted loans, the rest the bank managed to recover
df['Recovery_Amount'] = np.where(df['Default_Status'] == 'Yes', 
                                  df['loan_amount'] * 0.20, 0)
```

Step 10: Convert Data Types Properly

```python
df['year'] = df['year'].astype(int)
df['loan_amount'] = df['loan_amount'].astype(float)
df['income'] = df['income'].astype(float)
df['Credit_Score'] = df['Credit_Score'].astype(int)
df['term'] = df['term'].astype(int)
```

Step 11: Final Check

```python
print(df.isnull().sum().sum())  # should print 0 if all missing values are handled
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
problem_row = df[df['EMI'].isnull()]
print(problem_row[['loan_amount', 'rate_of_interest', 'term']])
```

If the interest rate is 0%, then logically, the EMI is simply the loan amount divided evenly across the number of months — no interest to add.

```python
# Check the exact row first (good practice, confirms before fixing)
print(df[df['EMI'].isnull()][['loan_amount', 'rate_of_interest', 'term']])

# Fix: if rate_of_interest is 0, EMI = loan_amount / term (simple division, no interest)
df.loc[df['EMI'].isnull(), 'EMI'] = df['loan_amount'] / df['term']

# Confirm the fix worked
print(df['EMI'].isnull().sum())
```
