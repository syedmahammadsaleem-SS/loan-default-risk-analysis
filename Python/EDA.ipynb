# Exploratory Data Analysis

```python
# Setup

import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("../Dataset/Cleaned_data/loan_data_cleaned.csv")
```

Chart 1: Loan Amount Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['loan_amount'], bins=30, color='steelblue', edgecolor='black')
plt.title('Distribution of Loan Amount')
plt.xlabel('Loan Amount')
plt.ylabel('Number of Loans')
plt.show()
```

Chart 2: Income Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['income'], bins=30, color='seagreen', edgecolor='black')
plt.title('Distribution of Monthly Income')
plt.xlabel('Income')
plt.ylabel('Number of Customers')
plt.show()
```

Chart 3: Credit Score Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['Credit_Score'], bins=30, color='darkorange', edgecolor='black')
plt.title('Distribution of Credit Score')
plt.xlabel('Credit Score')
plt.ylabel('Number of Customers')
plt.show()
```

Chart 4: Loan Purpose (Bar Chart)

```python
plt.figure(figsize=(10,5))
df['loan_purpose'].value_counts().plot(kind='bar', color='indianred')
plt.title('Number of Loans by Purpose')
plt.xlabel('Loan Purpose')
plt.ylabel('Number of Loans')
plt.xticks(rotation=0)
plt.show()
```

Chart 5: Employment / Business or Commercial (Bar Chart)

```python
plt.figure(figsize=(8,5))
df['business_or_commercial'].value_counts().plot(kind='bar', color='mediumpurple')
plt.title('Loans by Business or Commercial Type')
plt.xlabel('Type')
plt.ylabel('Number of Loans')
plt.xticks(rotation=0)
plt.show()
```

Chart 6: Loans by Branch (Bar Chart)

```python
plt.figure(figsize=(10,5))
df['Branch'].value_counts().sort_index().plot(kind='bar', color='teal')
plt.title('Number of Loans by Branch')
plt.xlabel('Branch')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45)
plt.show()
```

Chart 7: Loans by Region (Pie Chart)

```python
plt.figure(figsize=(7,7))
df['Region'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Loan Distribution by Region')
plt.ylabel('')
plt.show()
```

Chart 8: Default Status Overview (Pie Chart)

```python
plt.figure(figsize=(6,6))
df['Default_Status'].value_counts().plot(kind='pie', autopct='%1.1f%%', 
                                           colors=['lightgreen','salmon'], startangle=90)
plt.title('Loan Default vs Non-Default')
plt.ylabel('')
plt.show()
```

Chart 9: Interest Rate Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['rate_of_interest'], bins=30, color='goldenrod', edgecolor='black')
plt.title('Distribution of Interest Rate')
plt.xlabel('Interest Rate (%)')
plt.ylabel('Number of Loans')
plt.show()
```

Chart 10: Age Group Distribution (Bar Chart)

```python
plt.figure(figsize=(9,5))
df['age'].value_counts().sort_index().plot(kind='bar', color='cornflowerblue')
plt.title('Number of Loans by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Loans')
plt.xticks(rotation=0)
plt.show()
```

Chart 11: Gender Distribution (Bar Chart)

```python
plt.figure(figsize=(7,5))
df['Gender'].value_counts().plot(kind='bar', color='plum')
plt.title('Number of Loans by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Loans')
plt.xticks(rotation=0)
plt.show()
```

Bonus Chart 12: Default Rate by Loan Purpose (Connecting Default to a Category)

```python
default_by_purpose = df.groupby('loan_purpose')['Default_Status'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

plt.figure(figsize=(9,5))
default_by_purpose.plot(kind='bar', color='crimson')
plt.title('Default Rate (%) by Loan Purpose')
plt.xlabel('Loan Purpose')
plt.ylabel('Default Rate (%)')
plt.xticks(rotation=0)
plt.show()
```
