```python
import pandas as pd
import numpy as np

df = pd.read_csv('../Dataset/Cleaned_data/loan_data_cleaned.csv')
```

1. Mean (Average)

```python
print("Mean Loan Amount:", df['loan_amount'].mean())
print("Mean Income:", df['income'].mean())
print("Mean Credit Score:", df['Credit_Score'].mean())
```

2. Median (Middle Value)

```python
print("Median Loan Amount:", df['loan_amount'].median())
print("Median Income:", df['income'].median())
print("Median Credit Score:", df['Credit_Score'].median())
```

3. Mode (Most Common Value)

```python
print("Mode Loan Purpose:", df['loan_purpose'].mode()[0])
print("Mode Region:", df['Region'].mode()[0])
print("Mode Credit Score:", df['Credit_Score'].mode()[0])
```

4. Minimum and Maximum

```python
print("Min Loan Amount:", df['loan_amount'].min())
print("Max Loan Amount:", df['loan_amount'].max())
print("Min Credit Score:", df['Credit_Score'].min())
print("Max Credit Score:", df['Credit_Score'].max())
```

5. Variance

```python
print("Variance of Loan Amount:", df['loan_amount'].var())
```

6. Standard Deviation

```python
print("Std Dev of Loan Amount:", df['loan_amount'].std())
print("Std Dev of Credit Score:", df['Credit_Score'].std())
```

7. Quartiles

```python
print("25th Percentile (Q1) of Loan Amount:", df['loan_amount'].quantile(0.25))
print("50th Percentile (Q2/Median) of Loan Amount:", df['loan_amount'].quantile(0.50))
print("75th Percentile (Q3) of Loan Amount:", df['loan_amount'].quantile(0.75))
```

8. Percentiles (More Detailed Version of Quartiles)

```python
print("90th Percentile of Loan Amount:", df['loan_amount'].quantile(0.90))
print("95th Percentile of Loan Amount:", df['loan_amount'].quantile(0.95))
print("99th Percentile of Loan Amount:", df['loan_amount'].quantile(0.99))
```

9. Correlation

```python
correlation_matrix = df[['loan_amount', 'income', 'Credit_Score', 'rate_of_interest', 'LTV', 'dtir1']].corr()
print(correlation_matrix)
```

10. Correlation with Default Status (Business-Focused Bonus)

```python
df['Default_Flag'] = df['Default_Status'].map({'Yes': 1, 'No': 0})

correlation_with_default = df[['Default_Flag', 'loan_amount', 'income', 'Credit_Score', 
                                 'rate_of_interest', 'LTV', 'dtir1']].corr()['Default_Flag']
print(correlation_with_default)
```
