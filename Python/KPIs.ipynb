# KPIs (Key Performance Indicators)

```python
import pandas as pd

df = pd.read_csv('../Dataset/Cleaned_data/loan_data_cleaned.csv')
```

1. Total Customers

```python
total_customers = df['ID'].nunique()
print("Total Customers:", total_customers)
```

2. Total Loans

```python
total_loans = len(df)
print("Total Loans:", total_loans)
```

3. Approved Loans and Rejected Loans

```python
# All loans in our data are "approved" loans since they were issued
approved_loans = len(df)
print("Approved Loans (all issued loans):", approved_loans)
```

4. Default Loans

```python
default_loans = (df['Default_Status'] == 'Yes').sum()
print("Default Loans:", default_loans)
```

5. Default Rate

```python
default_rate = (default_loans / total_loans) * 100
print("Default Rate: {:.2f}%".format(default_rate))
```

6. Recovery Rate

```python
total_loss_potential = df['Loss_Amount'].sum() + df['Recovery_Amount'].sum()
total_recovered = df['Recovery_Amount'].sum()

recovery_rate = (total_recovered / total_loss_potential) * 100
print("Recovery Rate: {:.2f}%".format(recovery_rate))
```

7. Average Loan Amount

```python

avg_loan_amount = df['loan_amount'].mean()
print("Average Loan Amount: {:.2f}".format(avg_loan_amount))
```

8. Average Credit Score


```python
avg_credit_score = df['Credit_Score'].mean()
print("Average Credit Score: {:.2f}".format(avg_credit_score))
```

9. Average Income

```python
avg_income = df['income'].mean()
print("Average Income: {:.2f}".format(avg_income))
```

10. Total Loss

```python
total_loss = df['Loss_Amount'].sum()
print("Total Loss: {:.2f}".format(total_loss))
```

11. Total Recovery

```python
total_recovery = df['Recovery_Amount'].sum()
print("Total Recovery: {:.2f}".format(total_recovery))
```

Bringing All KPIs Together in One Clean Table

```python
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Customers', 'Total Loans', 'Approved Loans', 'Default Loans',
        'Default Rate (%)', 'Recovery Rate (%)', 'Average Loan Amount',
        'Average Credit Score', 'Average Income', 'Total Loss', 'Total Recovery'
    ],
    'Value': [
        total_customers, total_loans, approved_loans, default_loans,
        round(default_rate, 2), round(recovery_rate, 2), round(avg_loan_amount, 2),
        round(avg_credit_score, 2), round(avg_income, 2), round(total_loss, 2), round(total_recovery, 2)
    ]
})

print(kpi_summary)

kpi_summary.to_csv('../Reports/kpi_summary.csv', index=False)
print("KPI summary saved successfully!")
```
